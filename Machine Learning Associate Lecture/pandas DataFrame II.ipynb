```python
def package_category(package):
    if package < 300000:
        return "Bronze"
    elif 300000 <= package <= 600000:
        return "Silver"
    else:
        return "Gold"
```

```python
import numpy as np

def assign_category(department):
    if department is np.nan:
        return "Unassigned"
    else:
        return department

employee_info['department'] = employee_info['department'].apply(assign_category)
employee_info['department'].value_counts(normalize=True)
```

```python
## Change the null values in email column to "Unassigned"
```

```python
# Replace null values with average/median value. Imputing null values
median_salary = employee_info['salary'].median()
employee_info['salary'] = employee_info['salary'].fillna(median_salary)
```

```python
employee_info['salary'].dtype  # Numeric type
```

```python
employee_info['salary'] = employee_info['salary'].astype('str')  # String type
employee_info['salary']
```
