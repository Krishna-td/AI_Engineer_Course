# NumPy

NumPy stands for **Numerical Python**.

The core object of NumPy is **arrays**. The NumPy array is faster than Python's List object.

NumPy's array can store only one data type. Mixed data types are automatically
type-casted.

### Demo: Python List vs NumPy Array speed test

```python
list(range(1, 10))
```

```python
list_object = list(range(1, 1_00_001))  # Creating a list of 1,00,000 elements
len(list_object)
```

```python
%%timeit

double = []   # Multiplying every element in `list_object` by 2
for num in list_object:
    result = num * 2
    double.append(result)
```

```python
import numpy as np
```

```python
numpy_object = np.arange(1, 1_00_001)
len(numpy_object)
```

```python
%%timeit

numpy_object * 2
```

```python
[1, 2, 3] * 2
```

```python
x = np.array([1, 2, 3])
x * 2
```

```python
scores.min()  # Minimum value in the array
```

```python
scores.max()  # Maximum value in the array
```

```python
scores.sort()  # Sorts the values in the array in ascending order
```

```python
scores[::-1] # Sorts the values in descending order, but gives a new object
```

## Creating 2D arrays

```python
# Input: List of lists
temps = [[23.0, 25.0, 33.0], 
         [45.0, 28.0, 22.0],
         [33.5, 40.0, 22.5]]
```

```python
temps_array = np.array(temps)
temps_array
```

```python
temps_array.shape  # 3 rows and 3 columns
```

```python
temps_array.size  # Total number of elements: 9
```

```python
temps_array.ndim  # Number of dimensions: 2
```
