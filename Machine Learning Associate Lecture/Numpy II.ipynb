### Boolean Indexing

**Boolean Operators:**

* `&` - `and` operator
* `|` - `or` operator
* `~` - `not` operator

````python
grades = np.array([55, 40, 88, 99, 25, 10, 100])
````

````python
(grades > 80) & (grades < 100)
````

````python
boolean_array = grades > 50  # Boolean Array
boolean_array
````

````python
~boolean_array
````

````python
grades[boolean_array]
````
