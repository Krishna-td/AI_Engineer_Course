## Indexing in Series Object

Two types of indexing in `Series` object

1. Using `.loc[]` attribute
2. Using `.iloc[]` attribute

```python
products
```

```python
products.loc['TV'] = 65000  # Updating  a particular index
```

```python
products
```

```python
products.loc['Headphones']  # Label-based indexing
```

```python
products.iloc[3]  # Position-based indexing. Starts with 0.
```

```python
products.iloc[8]  # Out of range error
```

## Slicing in Series Object

```python
products
```

```python
products.loc['Laptop': 'TV']
```

Slicing can be used to update multiple indices using a single value.

```python
products.iloc[1:3] = np.nan
```

```python
products.iloc[2:4]
```

```python
products
```
