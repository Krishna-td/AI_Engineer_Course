The purpose of data visualization is to:

* Data Exploration
* Observe Patterns
* Extract useful information from data
* Find outliers
* Observe distribution of data

```python
# Import the libraries
import matplotlib.pyplot as plt  # Sub-package of matplotlib - pyplot
import seaborn as sns
```

```python
datasets = sns.get_dataset_names()  # Get available dataset names
datasets
```

```python
# Load the dataset
tips = sns.load_dataset('tips')
tips.head()  # First 5 rows of the dataset: tips
```

## Scatter Plot

Scatter plot is used to observe relationship/association between two continuous numerical variables.

```python
# Seaborn's scatter plot
fig, ax = plt.subplots(figsize=(14, 5))
sns.scatterplot(data=tips,
                x='total_bill',
                y='tip',
                # hue='time'   Use categorical column for `hue` parameter
               )
plt.xlabel("Total Bill")  # Change the x-axis label
plt.ylabel("Tip")  # Change the y-axis label
plt.title("Scatter Plot: Total Bill vs Tip", fontweight='bold', fontsize=14)  # Add title
```

```python
# Matplotlib's scatter plot
plt.scatter(x=tips['total_bill'], 
            y=tips['tip'])

plt.xlabel("Total Bill")
plt.ylabel("Tip")
```

## Distribution Plot

Distribution plot is used to check the distribution of a single continuous numerical variable.

It can be used to observe if a numerical variable follows a particular distribution (e.g. Normal distribution) and if skewness is present.

```python
iris = sns.load_dataset('iris')
iris.head()
```

```python
fig, ax = plt.subplots(figsize=(14, 5))
sns.kdeplot(data=iris, 
            x='petal_length',  # Kernel density estimate plot for "petal_length"
            hue='species', # Give categorical column for `hue` parameter
            fill=True)
```

```python
fig, ax = plt.subplots(figsize=(14, 5))
sns.scatterplot(data=iris,
                x='petal_length',
                y='petal_width',
                hue='species')
plt.title('Scatter Plot: Petal Length vs Petal Width', fontweight='bold')
```

## Box Plot

Box plot is used to observe the ranges of continuous numerical variables and also to detect
outliers.

```python
fig, ax = plt.subplots(figsize=(14, 5))
sns.boxplot(data=iris)  # Numeric variables are automatically extracted to create plot
```

## Bar Plot

Bar Plot is used to identify the association between a categorical variable and a numeric variable.

```python
fig, ax = plt.subplots(figsize=(14, 5))
sns.barplot(data=iris, 
            x='species',  # Categoric variable
            y='petal_width', # Numeric variable
            errorbar=None
           )
plt.title("Average Petal Length by Species", fontweight='bold', fontsize=14)
```

```python
fig, ax = plt.subplots(figsize=(14, 5))
sns.barplot(data=tips,
            x='size',
            y='tip',
            errorbar=None
           )
```

## Line Plot

Line plot is used to observe the trend in data. One of the variable should be a numeric variable. The other could be time data. Used commonly in stock market data.

```python
# dowjones
dowjones = sns.load_dataset('dowjones')
dowjones.tail()
```

```python
fig, ax = plt.subplots(figsize=(14, 5))
sns.lineplot(data=dowjones, 
             x='Date',  # x-axis
             y='Price',  # y-axis
             label='Dow Jones',
             color='green'
            )
ax.fill_between(x=dowjones['Date'], # x-range to fill 
                y1=dowjones['Price'],  # y-maximum
                y2=0, # y-minimum,
                alpha=0.1, # Increases the transparency of the color,
                color='green'
               )
plt.title("Stock Price of Dow Jones: 1914 - 1968", fontweight='bold', fontsize=14)
```
