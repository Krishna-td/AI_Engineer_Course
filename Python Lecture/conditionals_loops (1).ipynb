# Conditionals

Conditionals are used when we want to execute a piece of code based on a condition.

Examples:

* Password Authentication
* ATM Authenticate
* Supermarket Discount
* College Scholarship Eligibility

## Syntax

```python

    if <condition>:
        # python statements 
    elif <condition>:
        # python statements
    else:
        # python statements
```

````python
age = 17

if age >= 18:  # if False
    print("Eligible for Voting")
else:
    print("You're not eligible for Voting")
````

````python
income = 25000

if income <= 50000:
    print("Low income")
elif 50000 <= income <= 100000:
    print("Medium income")
else:
    print("High income")
````
