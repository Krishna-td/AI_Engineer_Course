## Dictionary Object

A dictionary object is used when we want to store relevant information together. In a dict
object, key-based indexing is used as opposed to integer based indexing in lists, tuples and strings.

```python
{'Kiran': [949223983, "kiran@yahoo.com"], "Vishal": [949239233, "vishal@gmail.com"]}
```

## Syntax

`{<key>:<value>, ...}` 

````python
students = {'Kiran': [949223983, "kiran@yahoo.com"], "Vishal": [949239233, "vishal@gmail.com"]}

students['Kiran']
````

## Non-empty dictionary

Rule for **keys** in dictionary:

1. The key must be an **immutable object**. One can use int, str, tuple as keys in a dict but not list.
2. Keys **should not contain duplicates**.

There are no rules for values in key-value pair.

````python
{1: 'a', 1: 'b'}
````

In dictionary, **slicing can't be done**

## Indexing

````python
personal_info = {"Kumar": 9493292393, "Rajesh": 73282399323, "Vignesh": 8839292393}
````

````python
# Extract Rajesh's phone number
personal_info["Rajesh"]
````

````python
personal_info_1 = {0: 9493292393, 1: 73282399323, 2: 8839292393}
personal_info_1[1]
````

### Update method

Unlike the update using the index method, we can update multiple key-value pairs at once.

````python
personal_info
````

````python
personal_info.update({"Kumar": 74839293333})
````

````python
personal_info
````

````python
personal_info.update({"Rajesh": 83839292939, "Gokul": 99923992323})
````

````python
personal_info
````

````python
personal_info.get("Gokul")  # almost same as using the index value
````

````python
personal_info['Ram']
````

````python
personal_info.get('Ram', "Not found")  # won't throw error if key isn't in dictionary
````

````python
personal_info.pop("Rajesh")
````

````python
personal_info
````

````python
personal_info.items()
````

````python
for student, number in personal_info.items():
    print("Student", student)
    print("Number", number)
    print("-----------")
````
