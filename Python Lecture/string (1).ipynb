## Indexing

In python, indexing is used to extract a single value in a object. Positive indexing starts with zero (`0`).

### Syntax
```python
object/variable[index_value]
```

Example:
**object**
```python
"python"[3]
```

**variable**
```python
course = "python"
course[3]
```

## Slicing

In Python, slicing is used to extract a range of values.

### Syntax
```python
variable/object[start_index:end_index:step_index]
```

````python
program = "Data Scientist"
````

````python
program[4:0:1]
````

````python
program[0:8:2]
````

````python
program[5:14:1]  # Positive indexing
````

````python
program[::]
````

````python
program[10:8:1]
````

* If step index is not given, it'll default to 1

If step index is _positive_:
* If start index is not given, it'll default to 0.
* If ends index is not given, it'll default to last index + 1

````python
program
````

````python
program[3:5:-1]  # Data, " a", 
````

````python
program[5:3:-1]
````

If step index is _negative_:

* start index should be larger than end index to get a non-empty string
* If start index is not given, it'll default to the last index
* If end index is not given, it'll default to first index - 1

````python
program
````

````python
program[-3:-5:-1]
````

````python
program[8::-1]
````

````python
program[:3:-1]
````
