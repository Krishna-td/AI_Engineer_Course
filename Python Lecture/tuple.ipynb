## Tuples

Tuples are very similar to lists but immutable objects. 

## Empty Tuple Object

```python
()
```

```python
type(())
```

## Non-empty Tuple Object

```python
(1, 'a', ('x', 'y', 'z'), 2.718, [1, 2])
```

## Single Element Tuple

```python
type(('a'))
```

```python
('a')
```

```python
('a',)
```

## Indexing and Slicing

Same as `list` and `str` objects.

## Methods

```python
dir(())
```

```python
dir(list())
```

## Membership Testing

Same as `list` and `str` objects.

## Functions

Same as `list` objects

## Mathematical Operators

```python
(1, 2, 3) + (4, 5, 6)
```

```python
(1, 2, 3) * 3
```

```python
(1, 2, 3) == (3, 2, 1)
```

```python
(1, 2, 3) == (1, 2, 3)
```

## Deleting a Tuple Object

```python
x = (1, 2, 3, 4, 5)
del x
```

## Multiple Assignment

```python
district, state, country = ("Coimbatore", "Tamil Nadu", "India")
print(district)
print(state)
print(country)
```

Tuples doesn't support **addition, updation, deletion of a single or slice of elements**.
