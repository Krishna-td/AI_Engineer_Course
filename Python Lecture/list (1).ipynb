## Updation in List

```python
students = ["Gowtham", "Ram", "Kumar", "Vishnu", "Joseph"]
students
```

```python
students[1] = "Vikas"  # Single update: indexing
```

```python
students
```

```python
students[3:] = ["Aishwarya", "Lakshmi"]  # multiple sequential updates: slicing
```

```python
students
```

```python
students[::2] = ["A", "B", "C"]
```

```python
students
```

```python
del students[::2]
```

```python
students
```
