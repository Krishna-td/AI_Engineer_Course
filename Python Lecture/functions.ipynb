## Functions

Functions are reusable block of code that can be run as many times as we want without duplicating the code.

## Syntax

```python
def <function_name>(<parameters>):
    # python statements
    return <expression>
```

````python
def greet(name):
    print("Hi", name)
    return name
````

````python
output = greet("Vijay") # "Vijay"
````

````python
print(output)
````

````python
def greet(name="Datamites"):  # Default arguments
    print("Hi", name)
    return name
````

````python
output = greet("Sathya")
````

````python
output
````

````python
def info(name, age, degree):
    print("Name", name)
    print("Age", age)
    print("Degree", degree)
````

````python
info(name="Vishnu", age=23, degree="BSc Chemistry")  # Key-word arguments
````

````python
info(name="Vishnu", 23, "BSc Chemistry")
````

````python
info("Vishnu", age=23, degree="BSc Chemistry")
````

````python
info_output = info("Vishnu", 23, "BSc Chemistry")  # Positional Arguments
````

````python
print(info_output)
````
