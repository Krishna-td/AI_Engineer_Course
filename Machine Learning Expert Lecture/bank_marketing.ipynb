# Classification of Term Deposit Subscribers

## Data Import 

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
bank_marketing = fetch_ucirepo(id=222) 
  
# data (as pandas dataframes) 
X = bank_marketing.data.features 
y = bank_marketing.data.targets 
```

## Exploratory Data Analysis

```python
type(X)
```

```python
type(y)
```

```python
import pandas as pd
```

```python
data = pd.concat([X, y], axis=1)
data.head()
```

```python
data.duplicated().any()  # No duplicate rows
```

```python
data.info()
```

```python
# Drop `poutcome` column due to large null values in the column
data = data.drop(columns='poutcome')
'poutcome' in data.columns
```

```python
# Imputing null values with most contacted type: cellular
most_contact_type = data['contact'].value_counts().idxmax()
most_contact_type
```

```python
data['contact'] = data['contact'].fillna(most_contact_type)
data['contact'].isnull().any()
```

```python
most_common_job = data['job'].value_counts().idxmax()
most_common_job
```

```python
data['job'] = data['job'].fillna(most_common_job)
data['job'].isnull().any()
```

```python
most_common_education = data['education'].value_counts().idxmax()
most_common_education
```

```python
data['education'] = data['education'].fillna(most_common_education)
data['education'].isnull().any()
```

```python
data.isnull().any()
```

```python
data['y'].value_counts(normalize=True)  # Class Imbalance
```

Most of the customers in the data are non-subscribers of the term deposit. **Class Imbalance exists
in our data**. The model has relatively little data to learn the characteristics of term deposit
subscribers.

This can be addressed by **undersampling** the majority class -- **non-subscribers** -- in our data
to maintain class balance.
