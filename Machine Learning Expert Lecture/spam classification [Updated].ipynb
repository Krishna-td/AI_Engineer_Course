# Spam Classification using K-Nearest Neighbors

## Data Import 

```python
from ucimlrepo import fetch_ucirepo 
  
# fetch dataset 
spambase = fetch_ucirepo(id=94) 
  
# data (as pandas dataframes) 
X = spambase.data.features 
y = spambase.data.targets 
```

```python
import pandas as pd

data = pd.concat([X, y], axis=1)
data.head()
```

## Train-test Split

```python
from sklearn.model_selection import train_test_split

target_column = "Class"
features = data.drop(columns=target_column)  # Every column except target column
target = data[target_column]
```

```python
X_train, X_test, y_train, y_test = train_test_split(features, 
                                                    target,
                                                    test_size=0.2,
                                                    stratify=target  # Maintain target distribution
                                                   )
```

## Model Training

```python
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=4)
```

```python
knn.fit(X_train, y_train)  # Training the model
```

## Model Evaluation

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score
```

```python
spam = data['Class'] == 1
len(data[spam])
```

## Train Accuracy

```python
y_pred_train = knn.predict(X_train)  # Model Predictions using features from train data
y_pred_train
```

```python
y_train  # True values of spam/non-spam from the dataset
```

```python
predictions = (y_train == y_pred_train)  # Comparing true values with model's prediction to find accuracy
predictions  # Boolean Series object
```

```python
predictions.sum()  # Get only the correct predictions - total count of `True`.
```

A prediction is correct if:

* model predicted as `spam` and true value is also `spam`
* model predicted as `non-spam` and true value is also `non-spam`

```python
2863/3368  # Computing accuracy manually. Out of 3368 emails given to the model, 2836 were predicted correctly
```

```python
accuracy_train = accuracy_score(y_true=y_train, y_pred=y_pred_train)
print(accuracy_train)
```

```python
precision_train = precision_score(y_true=y_train, y_pred=y_pred_train, pos_label=1)
print(precision_train)
```

```python
recall_train = recall_score(y_true=y_train, y_pred=y_pred_train, pos_label=1)
print(recall_train)
```

### Test Accuracy

```python
y_pred_test = knn.predict(X_test)  # Predict using test data's features
```

```python
accuracy_test = accuracy_score(y_true=y_test, y_pred=y_pred_test)
print(accuracy_test)
```

```python
precision_test = precision_score(y_true=y_test, y_pred=y_pred_test)
print(precision_test)
```

```python
recall_test = recall_score(y_true=y_test, y_pred=y_pred_test, pos_label=1)
print(recall_test)
```
